# Дерево решений: приёмная комиссияГотовые данные: enroll.csv. Выполняйте ячейки по порядку и сверяйте вывод со слайдами.

```python
import pandas as pdfrom sklearn.tree import DecisionTreeClassifier, export_text, plot_treeimport matplotlib.pyplot as plt
```

## Загружаем данныеОжидаем таблицу 12 строк × 4 столбца.

```python
df = pd.read_csv('enroll.csv')df
```

## Баланс классовСпрогнозируйте до запуска: сколько «да» и «нет»?

```python
df['enrolled'].value_counts()
```

## Обучаем деревоcriterion — критерий разбиения, max_depth — предел глубины.

```python
X = df.drop(columns=['enrolled'])y = df['enrolled']clf = DecisionTreeClassifier(criterion='entropy', max_depth=2, random_state=42)clf.fit(X, y)
```

## Читаем дерево в текстовом видеСоотнесите каждую строку вывода с элементом дерева: корень, узел, лист.

```python
print(export_text(clf, feature_names=list(X.columns)))
```

## Графическая визуализацияОтветьте в парах: какие классы смешаны в корне? В каком листе больше всего объектов? Какая ветка даёт «да» одним вопросом?

```python
plt.figure(figsize=(11, 5))plot_tree(clf, feature_names=list(X.columns), class_names=['нет', 'да'],          filled=True, rounded=True)plt.tight_layout()plt.show()
```

## Прогноз одного объектаСначала проведите объект по дереву ВРУЧНУЮ, затем запустите ячейку.

```python
clf.predict([[74, 1, 8]])
```

## Эксперимент с глубинойСпрогнозируйте точность глубины 1 до запуска.

```python
for depth in [1, 2, 3]:    m = DecisionTreeClassifier(criterion='entropy', max_depth=depth, random_state=42).fit(X, y)    print(depth, round(m.score(X, y), 2))
```
